```python
def classify_performance(marks):

    if marks < 10:
        return 'Low'

    elif marks < 15:
        return 'Medium'

    else:
        return 'High'


math_clean['performance'] = math_clean['G3'].apply(
    classify_performance
)

math_clean[['G3', 'performance']].head(10)
```

```python
performance_count = math_clean[
    'performance'
].value_counts()

display(performance_count)
```

```python
import os
import matplotlib.pyplot as plt
import seaborn as sns

# Create visualization directory
os.makedirs("../visualizations", exist_ok=True)

# Set professional visualization style
sns.set_theme(style="whitegrid", context="notebook")

print("Visualization environment ready!")
```

```python
plt.figure(figsize=(9, 6))

ax = sns.countplot(
    data=math_clean,
    x='result',
    order=['Pass', 'Fail']
)

plt.title(
    'Mathematics Student Pass/Fail Analysis',
    fontsize=16,
    fontweight='bold'
)

plt.xlabel('Academic Result', fontsize=12)
plt.ylabel('Number of Students', fontsize=12)

# Add count and percentage labels
total = len(math_clean)

for bar in ax.patches:
    count = int(bar.get_height())
    percentage = count / total * 100

    ax.annotate(
        f'{count}\n({percentage:.1f}%)',
        (bar.get_x() + bar.get_width() / 2, count),
        ha='center',
        va='bottom',
        fontsize=11
    )

plt.tight_layout()

plt.savefig(
    '../visualizations/pass_fail_analysis.png',
    dpi=300,
    bbox_inches='tight'
)

plt.show()
```
